# Detecção de Fraudes em Cartões de Crédito

## Introdução ao problema de negócio

As fraudes em transações com cartões de crédito representam um desafio para instituições financeiras, gerando prejuízos econômicos e reduzindo a confiança dos clientes. Com o aumento do volume de transações digitais, a identificação de atividades fraudulentas tornou-se cada vez mais importante.

Como utilizar técnicas de Machine Learning para identificar transações fraudulentas com cartão de crédito, reduzindo os prejuízos financeiros sem classificar incorretamente transações legítimas?

O objetivo deste projeto é desenvolver um modelo de classificação capaz de identificar padrões associados a fraudes, auxiliando na prevenção de perdas financeiras e na melhoria da segurança das transações.

## Etapas do Projeto

O desenvolvimento do projeto será dividido nas seguintes etapas:

1. **Análise do Desbalanceamento:** avaliar a proporção entre transações legítimas e fraudulentas.
2. **Preparação dos Dados:** separar as variáveis preditoras da variável-alvo e dividir os dados em conjuntos de treino e teste.
3. **Modelagem:** treinar algoritmos de Machine Learning para classificação de transações.
4. **Avaliação dos Modelos:** comparar os resultados utilizando métricas como Precision, Recall, F1-Score e ROC-AUC.
5. **Conclusão:** interpretar os resultados e avaliar a aplicabilidade do modelo ao problema de negócio.
